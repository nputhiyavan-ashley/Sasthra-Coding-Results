```python
s = input()
stack = []
opening = {'(', '[', '{'}
matching = {
    ')': '(',
    ']': '[',
    '}': '{'
}
for i, ch in enumerate(s):
    if ch in opening:
        stack.append((ch, i))
    elif ch in matching:
        if not stack:
            print("ERROR:" + str(i))
            break
        if stack[-1][0] != matching[ch]:
            print("ERROR:" + str(i))
            break
        stack.pop()
else:
    if stack:
        print("ERROR:" + str(stack[0][1]))
    else:
        print("BALANCED")
```
