```python
def norm(data_str):
    inp = data_str.strip().split()
    if not inp:
        return

    r = int(inp[0])
    c = int(inp[1])

    idx = 2
    matrix = []
    for _ in range(r):
        row = [float(x) for x in inp[idx:idx + c]]
        matrix.append(row)
        idx += c


    cm = []
    cr = []
    for j in range(c):
        col_values = [matrix[i][j] for i in range(r)]
        min_val = min(col_values)
        max_val = max(col_values)
        cm.append(min_val)
        cr.append(max_val - min_val)

    output = []
    for i in range(r):
        row_scaled = []
        for j in range(c):
            if cr[j] == 0:
                row_scaled.append(f"{0.0:.4f}")
            else:
                scaled = (matrix[i][j] - cm[j]) / cr[j]
                row_scaled.append(f"{scaled:.4f}")
        output.append(" ".join(row_scaled))

    print("\n".join(output))


test_input_1 = """
3 2
10 5
20 5
30 5
"""
norm(test_input_1)

test_input_2 = """
4 3
-10 0 100
0 50 200
10 100 300
20 150 400
"""
norm(test_input_2)
```
