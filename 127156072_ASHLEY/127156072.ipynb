```python
n,m=map(int,input().split())
graph={}
count=0
for i in range(n):
    graph[i+1]=1
for _ in range(m):
    u,v=map(int,input().split())
    graph[u]=graph.get(u,1)+1
for node,freq in graph.items():
    if freq>1:
        count+=1
print(count)
edges=[]
for key,value in graph.items():
    edges.append(value)
print(*edges)
```
