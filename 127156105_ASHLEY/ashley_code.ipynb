```python
n,target=map(int,input().split())
numbers=list(map(int,input().split()))
first_position={}
for i in range(n):
    value=numbers[i]
    required=target-value
    if required in first_position:
        print(first_position[required],i)
        break
    if value not in first_position:
        first_position[value]=i
else:
    print("NONE")
    
```
