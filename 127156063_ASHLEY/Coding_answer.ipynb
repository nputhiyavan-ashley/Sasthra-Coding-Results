```python
'''
    this is the input for the number of events in different timestamps
'''
n = int(input())

events = {}

''' 
    loop throught the number of inputs to get the events
'''

for _ in range(n):
    line = input()

    timestamp, event = line.split(',')

    date = timestamp.split('T')[0]

    key = (date, event)

    events[key] = events.get(key, 0) + 1


for (date, event) in sorted(events):
    print(f"{date},{event},{events[(date, event)]}")
```
